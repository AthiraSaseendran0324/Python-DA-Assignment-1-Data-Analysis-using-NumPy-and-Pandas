##Numpy Array Operations:
Scenario: Here given a list of daily average temperatures (in degrees Celsius)
recorded over two weeks. Your task is to use NumPy to analyze this data.

####1. Create a 1D NumPy Array:
Create a 1D NumPy array named temperatures_w1 for Week 1 with the
following values:
[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9]

```python
import numpy as np
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
```

First, I imported the NumPy library using import numpy as np. NumPy is a Python library used for numerical calculations and working with arrays.

Next, I created a one dimensional NumPy array named temperatures_w1 using the np.array() function. then added values [22.5 25.3 20.8 23.4 26.1 24.8 21.9] in it. then using print(), displayed the values.

###2. Inspection and Properties:
● Inspect the shape, data type, and the number of elements in the array.
Inspect the shape of the temperatures_w1 array.

```python
print("Shape:",temperatures_w1.shape)
print("Data type:",temperatures_w1.dtype)
print("number of elements:",temperatures_w1.size)
```

In this task, I inspected the properties of the temperatures_w1 NumPy array using its built-in attributes.

The shape attribute returns the dimensions of the array. The output (7,) indicates that the array contains seven elements in one dimension.

The dtype attribute returns the data type of the elements. Here, float64 indicates that the temperature values are stored as floating point numbers.

The size attribute returns the total number of elements in the array. Since there are seven temperature values, the output is 7.


###3. Array Operations:
● Convert the temperatures_w1 from Celsius to Fahrenheit using the
formula: Fahrenheit = (Celsius * 9/5) + 32

● Find the maximum, minimum, and mean temperatures for the week.

```python
temperatures_f = (temperatures_w1 * 9/5) + 32
print("Temperatures in Fahrenheit:", temperatures_f)
```

I converted the temperatures from Celsius to Fahrenheit using the given formula. I multiplied the temperatures_w1 array by 9/5 and then added 32.
I stored the converted values in a new array named temperatures_f and used the print() function to display the temperatures in Fahrenheit.

```python
print("Maximum temperature:",np.max(temperatures_w1))
print("Minimum temperature:",np.min(temperatures_w1))
print("Mean temperature:",np.mean(temperatures_w1))
```

 I used NumPy functions to calculate the maximum, minimum, and mean temperatures for Week 1.

np.max(temperatures_w1) returns the maximum temperature in the array.

np.min(temperatures_w1) returns the Minimum temperature in the array.

np.mean(temperatures_w1) calculates the mean (average ) of all 7 temperature values.

###4. Array Slicing and Indexing:
● Extract the temperatures for the first three days of the week.

● Extract the temperatures for the weekend (last two days).

● Extract the temperatures for the middle three days of the week.

```python
print(temperatures_w1[0:3])
print(temperatures_w1[-2:])
print(temperatures_w1[2:5])
```

In this task, I used array slicing to extract specific temperature values from the temperatures_w1 NumPy array.

temperatures_w1[0:3] extracts the temperatures for the first three days. It starts from index 0 and stops before index 3(index 2).

temperatures_w1[-2:] extracts the temperatures for last two days. Negative indexing allows us to access elements from the end of the array. The index -2 represents the second-last element, and : selects all elements from that position to the end.

temperatures_w1[2:5] extracts the temperatures for the middle three days. It starts from index 2 and stops before index 5.

###5. Create a 2D Array:
Create a 2D NumPy array named temperatures where each row represents
temperatures for a different week:

★ Row 1: [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9] (Week 1)

★ Row 2: [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1] (Week 2)

The array contains two rows. The first row represents the daily temperatures for Week 1, and the second row represents the daily temperatures for Week 2. Each row contains seven temperature values, one for each day of the week.



###6. Inspect and Slice the 2D Array:
● Inspect and print the shape, data type, and total number of elements in the
temperatures array.

● Extract the temperatures for each week and the weekends (last two days)
for both weeks.

```python
print("Shape:",temperatures.shape)
print("Data type:",temperatures.dtype)
print("Total number of elements:",temperatures.size)

print("Week 1 temperatures:",temperatures[0])
print("Week 2 temperatures:",temperatures[1])

print("Week 1 weekend temperatures:",temperatures[0,-2:])
print("Week 2 weekend temperatures:",temperatures[1,-2:])
```


The shape attribute returns (2, 7), which means the array contains two rows and seven columns.

The dtype attribute displays the data type of the temperature values, which is float64.

The size attribute returns the total number of elements in the array. Since there are two weeks with seven temperatures each, the total is 14.

temperatures[0] extracts the first row, which contains Week 1 temperatures.

temperatures[1] extracts the second row, which contains Week 2 temperatures.

temperatures[0, -2:] extracts the last two temperatures from the first row,Week 1's weekend.

temperatures[1, -2:] extracts the last two temperatures from the second row,Week 2's weekend.

##Pandas Series:

###1. Creating Pandas Series

● Create a Pandas Series named marks with the following data:

➢ Marks: 95, 92, 89, 85, 80

➢ Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'

```python
import pandas as pd
marks = pd.Series([95, 92, 89, 85, 80],index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'],name="Marks")
print(marks)
```

First, I imported the Pandas library using import pandas as pd. Pandas is a Python library used for working with and analyzing structured data.
Next, I created a Series named marks using the pd.Series() function. I added the five marks as the data and assigned custom index labels to each mark using the index parameter(named indices as 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5)

###2. Indexing and Slicing:
● Use the integer index position to access the mark of the 1st rank student.

● Use the loc accessor to retrieve the marks of the top 3 ranks by
specifying their index labels.

● Use the iloc accessor to retrieve the mark of the 3rd rank student.

● Apply a boolean mask to filter the Series and retrieve the ranks where the
marks are greater than 90.

```python
print(marks.iloc[0])
```

```python
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])
```

```python
print(marks.iloc[2])
```

```python
print(marks[marks>90])
```

 1. used the iloc accessor to access the mark by its integer position. Since ython indexing starts at 0, marks.iloc[0] returns the first student's mark, which is 95

 2. used the loc accessor to retrieve marks using the index labels Rank1, Rank2, and Rank3. It returns the marks corresponding to these labels.
 3. used marks.iloc[2] to retrieve the third student's mark. Since indexing starts at 0, position 2 represents the third element, which has a mark of 89.
 4.  used Boolean masking with the condition "marks>90" to select only the students whose marks are greater than 90. The output displays Rank1 with 95 marks and Rank2 with 92 marks.

###3. Manipulating Series
● Modify the mark of the student with the 1st rank to 100.

● Remove the entry corresponding to the last rank from the Series.

● Compute the CGPA by dividing each mark in the Series by 10.

```python
marks['Rank1']=100
print(marks)
```

```python
marks = marks.drop('Rank5')
print(marks)
```

```python
cgpa = marks/10
print(cgpa)
```

1. Used the index label 'Rank1' to modify the student's mark to 100 in the marks Series. The updated value is assigned using marks['Rank1'] = 100.
2. Used the drop() method to remove the entry with the index label Rank5 from the marks Series.
3. Divided each value in the marks Series by 10 to calculate the CGPA values.I stored the results in a new Series named cgpa and printed it.

###3. Manipulating the DataFrame:
● Modify the 'Amount' for the transaction with TransactionID 102 to 165.

● Add a new column 'Discount' by calculating 10% of the 'Amount'.

● Remove the row with TransactionID 109.

● Delete the 'Discount' column from the DataFrame.

- Modify the 'Amount' for the transaction with TransactionID 102 to 165.

```python
transactions.loc[transactions['TransactionID']==102,'Amount']=165
print(transactions)
```

Used the loc accessor to identify the row where TransactionID equals 102 and selected its Amount column. I then updated the amount to 165 and printed the modified DataFrame.

- Add a new column 'Discount' by calculating 10% of the 'Amount'.

```python
transactions['Discount']=transactions['Amount']*0.10
print(transactions)
```

Created a new column named Discount by multiplying each value in the Amount column by 0.10. This calculates 10% of the transaction amount for every row. The new column is added to the existing DataFrame.

- Remove the row with TransactionID 109.

```python
transactions = transactions.drop(transactions[transactions['TransactionID']==109].index)
print(transactions)
```

Identified the index of the row where TransactionID equals 109 and used the drop() method to remove that row. Assigned the result back to transactions to save the updated DataFrame.

- Delete the 'Discount' column from the DataFrame.

```python
transactions = transactions.drop('Discount', axis=1)
print(transactions)
```

Used the drop() method with axis=1 to remove the Discount column from the DataFrame. I assigned the result back to transactions to save the changes.
